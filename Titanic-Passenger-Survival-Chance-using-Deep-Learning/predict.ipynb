```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import tensorflow as tf
```

```python
with open('models/label_encoder.pkl', 'rb') as file:
    label = pickle.load(file)
```

```python
with open('models/scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
```

```python
with open('models/onehot_encoder.pkl', 'rb') as file:
    onehot = pickle.load(file)
```

```python
model = load_model('models/model.h5')
```

```python
user = {'Pclass' : 3,
        'Sex' : 'female', 
        'SibSp' : 1, 
        'Parch' : 1, 
        'Fare' : 11,
       'Embarked' : 'Queenstown'
}
```

```python
user = pd.DataFrame([user])
user
```

```python
user['Sex'] = label.transform(user['Sex'])
embarked = onehot.transform(user[['Embarked']])
embarked = pd.DataFrame(embarked,columns=onehot.get_feature_names_out())
user = pd.concat([user.drop(columns=["Embarked"]), embarked], axis=1)
user
```

```python
user[['Pclass', 'SibSp', 'Parch', 'Fare']] = scaler.transform(user[['Pclass', 'SibSp', 'Parch', 'Fare']])
user
```

```python
y = model.predict(user)
```

```python
y
```

```python
y = y[0][0]
```

```python
if y>0.5:
    print('Person survived')
else:
    print("Person doesn't survived")
```
